### 1. Найти сумму четных цифр введенного натурального числа

```python
number = input("Введите натуральное число: ")
```

```python
even_sum = 0
for digit in number:
    if digit.isdigit() and int(digit) % 2 == 0:
        even_sum += int(digit)
```

```python
print(f"Сумма четных цифр: {even_sum}")
```

### 5. Реализуйте программу «Ювелирный магазин», которая будет включать в себя шесть пунктов меню. 
У вас есть словарь, где ключ – название изделия. 
Значение – список, который содержит состав изделия(золото, серебро, и т.п.), цену и кол-во (шт),которое есть в магазине.

5. Покупка

    В пункте «Покупка» необходимо совершить покупку, с

программы. Посчитать цену выбранных товаров и сколько товаров

6. До свидания

```python
shop = {
    "Кольцо":  ["золото", 5000, 10],
    "Серьги":  ["серебро", 3000, 15],
    "Браслет": ["золото, серебро", 8000, 5],
    "Цепочка": ["золото", 12000, 7],
    "Кулон":   ["серебро", 2500, 20],
    "Брошь":   ["золото, платина", 15000, 3]
}
```

```python
while True:
    print("1. Просмотр описания")
    print("2. Просмотр цены")
    print("3. Просмотр количества")
    print("4. Вся информация")
    print("5. Покупка")
    print("6. Выход")

    choice = input("Выберите пункт: ")

    if choice == "1":
        for name, info in shop.items():
            print(f"{name} – {info[0]}")
    elif choice == "2":
        for name, info in shop.items():
            print(f"{name} – {info[1]} руб.")
    elif choice == "3":
        for name, info in shop.items():
            print(f"{name} – {info[2]} шт.")
    elif choice == "4":
        for name, info in shop.items():
            print(f"{name} – состав: {info[0]}, цена: {info[1]} руб., кол-во: {info[2]} шт.")
    elif choice == "5":
        product = input("Введите название изделия: ")
        if product in shop:
            amount = int(input("Введите количество: "))
            if amount <= shop[product][2]:
                total = amount * shop[product][1]
                shop[product][2] -= amount
                print(f"Куплено: {product} x {amount}")
                print(f"Стоимость покупки: {total} руб.")
                print(f"Осталось на складе: {shop[product][2]} шт.")
            else:
                print(f"Недостаточно товара. В наличии только {shop[product][2]} шт.")
        else:
            print("Такого изделия нет в магазине")
    elif choice == "6":
        print("До свидания")
        break
    else:
        print("Неверный пункт меню")
```

### 1. Написать функцию is_prime, принимающую 1 аргумент — число от 0 до 1000, и возвращающую True, если оно простое, и False - иначе. 

```python
def is_prime(n):
    if not isinstance(n, int):
        raise TypeError("Аргумент должен быть целым числом")
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
```

```python
while True:
    try:
        num = int(input("Введите число от 0 до 1000: "))
        if not (0 <= num <= 1000):
            raise ValueError("Число должно быть в диапазоне 0..1000")
        result = is_prime(num)
        print(f"Число {num} простое? {result}")
        break
    except ValueError as e:
        print("Ошибка:", e)
    finally:
        print("The end")
        
```

### 3. Дана целочисленная прямоугольная матрица. Определить количество столбцов, не содержащих ни одного нулевого элемента.

```python
def count_cols(matrix):
    return sum(all(row[j] != 0 for row in matrix) for j in range(len(matrix[0])))

matrix = [
    [0, 0, 3],
    [7, 1, 2],
    [0, 8, 3]
]
try:
    print("Матрица:")
    for row in matrix:
        print(row)
    print("Столбцов без нулей:", count_cols(matrix))
except Exception as e:
    print("Ошибка:", e)
finally:
    print("Готово")
```

### 2. Напишите функцию, которая будет принимать один аргумент. 
* Если в функцию передаётся список, найти среднее арифметическое все его чисел.
* Если словарь, Отсортируйте по значению в порядке возрастания.
* Число – Вывести в обратном порядке.
* Строка – найти количество слов. Вывести самое длинное слово.
* Сделать проверку со всеми этими случаями.

```python
def process(data):
    if isinstance(data, list):
        return sum(data) / len(data)
    elif isinstance(data, dict):
        return dict(sorted(data.items(), key=lambda x: x[1]))
    elif isinstance(data, (int, float)):
        return str(data)[::-1]
    elif isinstance(data, str):
        words = data.split()
        return len(words), max(words, key=len)
    else:
        raise TypeError("Неподдерживаемый тип данных")
```

```python
cases = [
    [1, 2, 3, 4, 5],
    {"a": 3, "b": 1, "c": 2},
    12345,
    "мама и папа вместе навсегдаа"
]
```

```python
for c in cases:
    try:
        print(f"Вход: {c!r}")
        print("Результат:", process(c))
    except Exception as e:
        print("Ошибка:", e)
    finally:
        print("Готово\n")
```

## Задание 1: Калькулятор. Создайте класс, где реализованы функции(методы) математических операций. А также функция, для ввода данных. 

```python
class Calculator:
    def add(self, a, b):
        return a + b
    def sub(self, a, b):
        return a - b
    def mul(self, a, b):
        return a * b
    def div(self, a, b):
        if b == 0:
            raise ZeroDivisionError("Деление на ноль невозможно")
        return a / b
    def get_data(self):
        a = float(input("Введите первое число: "))
        b = float(input("Введите второе число: "))
        op = input("Введите операцию (+, -, *, /): ")
        return a, b, op

calc = Calculator()

try:
    a, b, op = calc.get_data()
    if op == "+":
        print("Результат:", calc.add(a, b))
    elif op == "-":
        print("Результат:", calc.sub(a, b))
    elif op == "*":
        print("Результат:", calc.mul(a, b))
    elif op == "/":
        print("Результат:", calc.div(a, b))
    else:
        print("Неизвестная операция")
except ZeroDivisionError as e:
    print("Ошибка:", e)
except ValueError:
    print("Ошибка: введено не число")
finally:
    print("Работа завершена")
```

## Задание 2:
Базовый класс Worker (работник).
1. Определить атрибуты: name, surname, position (должность), income
(доход);
2. Последний атрибут должен быть защищённым и ссылаться на словарь,
содержащий элементы: оклад и премия, например, {"wage": wage, "bonus":
bonus};
3. Создать класс Position (должность) на базе класса Worker;

```python
class Worker:
    def __init__(self, name, surname, position, wage, bonus):
        self.name = name
        self.surname = surname
        self.position = position
        self._income = {"wage": wage, "bonus": bonus}


class Position(Worker):
    def get_full_name(self):
        return f"{self.name} {self.surname}"

    def get_total_income(self):
        return self._income["wage"] + self._income["bonus"]

p1 = Position("Мария", "Марияновна", "Менеджер", 50000, 15000)
p2 = Position("Мария", "Петрова", "Бухгалтер", 45000, 10000)

for p in (p1, p2):
    print("Имя:", p.name)
    print("Фамилия:", p.surname)
    print("Должность:", p.position)
    print("Доход (словарь):", p._income)
    print("Полное имя:", p.get_full_name())
    print("Общий доход:", p.get_total_income())
```
